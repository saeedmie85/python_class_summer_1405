```python
with open('scores.txt','w') as f:
    f.write('name, score1, score2\n')
    for _ in range(5):
        name = input('name?')
        score1 = input('score1?')
        score2 = input('score2?')
        f.write(f'{name}, {score1}, {score2}\n')
```

```python
with open('scores.tsv','w') as f:
    f.write('name\t score1\t score2\n')
    for _ in range(5):
        name = input('name?')
        score1 = input('score1?')
        score2 = input('score2?')
        f.write(f'{name}\t {score1}\t {score2}\n')
```
