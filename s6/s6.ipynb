```python
d1 = {'a': 100, 'b': 200, 'c':300}
d2 = {'a': 300, 'b': 200, 'd':400}
result = d2.copy()
for k in d1:
    if k in d2:
        result[k] += d1[k]
    else:
        result[k] = d1[k]
result
```
