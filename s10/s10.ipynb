```python
class Ball:
    """
    Short description of the class.

    Attributes:
        attribute1 (type): Description of the attribute.
        attribute2 (type): Description of the attribute.
    """
    def __init__(self,color:str, number:str, x:float, y:float)->None:
        self.color = color
        self.number = number
        self.move(x,y)

    def move(self, x, y):
        self.x = x
        self.y = y

    def reset(self):
        self.move(0,0)

    def dist(self, other:Ball)->float:
        """
        Calculate the distance of a ball with other ball.

        Args:
            Other (Ball): A ball object from Ball class.

        Returns:
            float: the distance of two balls.
        """
        return ((self.x - other.x)**2 + (self.y - other.y)**2)**.5

    def is_collide(self, other:Ball)->bool:
        return self.dist(other) <= .5


    def __repr__(self):
        return f'Ball Num:{self.number}, Ball color:{self.color}'
    
```

```python
b1 = Ball('red','1',21.5,15.6)
b2 = Ball('blue','2',25.5,35.6)
b3 = Ball('yelow', '3', 26, 36)
```

```python
class Contact:
    CONTACT_COUNT = 0

    def __init__(self,name, phone):
        self.name = name
        self.phone = phone
        Contact.CONTACT_COUNT += 1

    def call(self):
        print(f'calling to {self.phone}... hi {self.name}')
        print('tanx')

    def __repr__(self):
        return f'{self.__class__.__name__}:{self.name}'


class Suplier(Contact):

    def send_order(self):
        print('send order')

class Freind(Contact):

    def send_sms(self, messege):
        print(f'send sms to {self.name} messege: {messege}')

    def call(self):
        print('hi how are you')
        super().call()

class Uk(Contact):

    def call(self):
        self.name = input('enter your name')
        super().call()
```

```python
c1 = Contact('hasan',323123245)
s1 = Suplier('tahlildadeh',2134152452)
f1 = Freind('ali',5454545)
uk1 = Uk('',454545)
```
