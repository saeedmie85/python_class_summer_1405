```python
w = float(input('please enter your weight:'))
h = float(input('please enter your height:'))
bmi = w / (h ** 2)
print(bmi)
if bmi < 16:
    print('Sever Thinnes')
elif 16 <= bmi < 17:
     print('Moderate Thinness')
elif bmi >= 17 and bmi < 18.5:
     print('Mild Thinness')
else:
     print('normal')
```

```python
score = float(input('Enter your grade:'))
if 0 <= score <= 1:
    if score >= 0.9:
        print('A')
    elif score >= .8:
        print('B')
    elif score >= .7:
        print('C')
    elif score >= .6:
        print('D')
    else:
        print('F')
else:
    print('Wrong entry')
```
