4.	هر ۵ ثانیه مسیر را چک کنید و اگر فایل جدید اضافه شد یا حذف شد، پیام بدهید.

```python
import time
from datetime import datetime

while True:
    before_set = set(os.listdir(r'exc2\d_0'))
    time.sleep(5)
    after_set = set(os.listdir(r'exc2\d_0'))
    new_files = (after_set - before_set).union(
        before_set - after_set
    )
    if len(new_files) >= 1:
        changed = 'added' if len(after_set) > len(before_set) else 'removed'
        print(f'{datetime.now().strftime('%M:%S')} : {new_files} {changed}')
```

```python
import shutil

image_extensions = [
    ".jpg",
    ".jpeg",
    ".png",
    ".gif",
    ".bmp",
    ".webp",
    ".tiff",
    ".tif",
    ".svg",
    ".ico",
    ".avif",
    ".heic",
    ".heif",
    ".jfif",
    ".apng"
]
# Audio files
audio_extensions = [
    ".mp3",
    ".wav",
    ".aac",
    ".flac",
    ".ogg",
    ".oga",
    ".m4a",
    ".wma",
    ".aiff",
    ".aif",
    ".alac",
    ".opus",
    ".amr",
    ".mid",
    ".midi",
    ".ape",
    ".ac3",
    ".caf"
]

# Video files
video_extensions = [
    ".mp4",
    ".mkv",
    ".avi",
    ".mov",
    ".wmv",
    ".flv",
    ".webm",
    ".mpeg",
    ".mpg",
    ".m4v",
    ".3gp",
    ".3g2",
    ".ts",
    ".mts",
    ".m2ts",
    ".vob",
    ".ogv",
    ".asf",
    ".rm",
    ".rmvb",
    ".divx"
]

# Document files
document_extensions = [
    ".pdf",
    ".doc",
    ".docx",
    ".docm",
    ".odt",
    ".rtf",
    ".txt",
    ".md",
    ".tex",
    ".pages",
    ".xls",
    ".xlsx",
    ".xlsm",
    ".csv",
    ".ods",
    ".ppt",
    ".pptx",
    ".pptm",
    ".odp",
    ".epub",
    ".mobi",
    ".azw",
    ".azw3",
    ".html",
    ".htm"
]

root = 'exc2'

if not os.path.isdir(root):
    print("Folder does not exist!")

else:
    folders = ["image", "audio", "video", "document"]

    # Create destination folders
    for folder in folders:
        os.makedirs(os.path.join(root, folder),
                     exist_ok=True)

    # Check files
    for filename in os.listdir(root):

        source = os.path.join(root, filename)

        if not os.path.isfile(source):
            continue

        extension = os.path.splitext(filename)[1].lower()

        if extension in image_extensions:
            folder = "image"

        elif extension in audio_extensions:
            folder = "audio"

        elif extension in video_extensions:
            folder = "video"

        elif extension in document_extensions:
            folder = "document"

        else:
            continue

        destination = os.path.join(root, folder, filename)

        # Prevent overwriting existing files
        if os.path.exists(destination):
            print(f"Skipped (already exists): {filename}")
            continue

        shutil.move(source, destination)

        print(f"{filename} moved to {folder}")

    print("Sorting completed!")
```
