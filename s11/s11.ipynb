```python
from IPython.display import Audio, display
class IceCream:
    CAPACITY = 200000
    INVENTORY = 0

    def __init__(self, price, weight, number):
        self.price = price
        self.weight = weight
        self.number = number
        IceCream.INVENTORY += number * weight

    def sell(self, num):
        if num <= self.number:
            self.number -= num
            IceCream.INVENTORY -= num * self.weight
            return num
        return 'There is no enough icecream'

    def buy(self, num):
        new_weight = num * self.weight + IceCream.INVENTORY 
        if new_weight <= IceCream.CAPACITY:
            self.number += num
            IceCream.INVENTORY = new_weight
            return new_weight
        return "There is no capacity"

class MusicalIceCream(IceCream):

    def __init__(self, price, weight, number, music_pth):
        super().__init__(price, weight, number)
        self.music = music_pth

    def sell(self, num):
        super().sell(2 * num)

    def play_music(self):
        display(Audio(filename=self.music, autoplay=True))

```
